Approach for scraping from dexscreener

- using Apify

https://apify.com/muhammetakkurtt/dexscreener-scraper

Hit "Run" and wait for a min. You'll then get top 100 trending tokens. Copy the link to the dataset/download it

```python
import pandas as pd
```

```python
dataset_id = "pDDPgMDNve6I966du"
```

```python
url = f"https://api.apify.com/v2/datasets/{dataset_id}/items?clean=true"
```

```python
df = pd.read_json(url)
```

```python
df
```

```python
df.columns
```

```python
df.head()
```

```python
# First, expand the nested dicts
df['liquidity_usd'] = df['liquidity'].apply(lambda x: x.get('usd') if isinstance(x, dict) else None)
df['volume_h24'] = df['volume'].apply(lambda x: x.get('h24') if isinstance(x, dict) else None)

# Now compute the ratios
df['liquidity_to_volume'] = df['liquidity_usd'] / df['volume_h24']
df['fdv_to_liquidity'] = df['fdv'] / df['liquidity_usd']
```
